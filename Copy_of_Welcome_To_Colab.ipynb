```
# ================================================================
# 🌾 AI-BASED CROP YIELD PREDICTION SYSTEM
# ================================================================
# Complete Google Colab Code
#
# Model       : CatBoost Regressor
# Target      : Yield
# Dataset     : crop_yield.csv
#
# Outputs:
#   - R² Score
#   - R² Percentage
#   - MAE
#   - RMSE
#   - MAPE
#   - SMAPE
#   - Accuracy
#   - Precision
#   - Recall
#   - F1 Score
#   - Confusion Matrix
#   - Feature Importance
#   - Actual vs Predicted graph
#   - Residual graph
#   - Model comparison
#   - Cross-validation
#   - Trained model
#   - Prediction CSV
#   - Complete results ZIP
# ================================================================


# ================================================================
# 1. INSTALL REQUIRED LIBRARIES
# ================================================================

!pip install -q catboost


# ================================================================
# 2. IMPORT LIBRARIES
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import json
import zipfile
import shutil
import warnings

warnings.filterwarnings("ignore")

from google.colab import files

from sklearn.model_selection import train_test_split, KFold
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

from sklearn.preprocessing import OrdinalEncoder

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)

from catboost import CatBoostRegressor


print("=" * 70)
print("🌾 AI-BASED CROP YIELD PREDICTION SYSTEM")
print("=" * 70)
print("All libraries imported successfully!")


# ================================================================
# 3. UPLOAD CSV FILE
# ================================================================

print("\nPlease upload your crop_yield.csv file.")

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

print("\nUploaded file:", file_name)


# ================================================================
# 4. LOAD DATASET
# ================================================================

df = pd.read_csv(file_name)

print("\nDataset loaded successfully!")

print("Dataset shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())


# ================================================================
# 5. BASIC DATASET INFORMATION
# ================================================================

print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

print("\nNumber of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())


# ================================================================
# 6. REMOVE DUPLICATES
# ================================================================

df = df.drop_duplicates().reset_index(drop=True)

print("\nDataset after removing duplicates:", df.shape)


# ================================================================
# 7. DEFINE TARGET
# ================================================================

TARGET = "Yield"

if TARGET not in df.columns:
    raise ValueError(
        "ERROR: 'Yield' column was not found in the dataset."
    )

print("\nTarget variable:", TARGET)


# ================================================================
# 8. CLEAN TARGET
# ================================================================

df[TARGET] = pd.to_numeric(
    df[TARGET],
    errors="coerce"
)

df = df.dropna(
    subset=[TARGET]
).reset_index(drop=True)

print(
    "\nDataset after cleaning target:",
    df.shape
)


# ================================================================
# 9. DESCRIPTIVE STATISTICS
# ================================================================

print("\n" + "=" * 70)
print("DESCRIPTIVE STATISTICS")
print("=" * 70)

display(
    df.describe(include="all")
)


# ================================================================
# 10. EXPLORATORY DATA ANALYSIS
# ================================================================

print("\nGenerating EDA graphs...")


# Yield Distribution
plt.figure(figsize=(10, 6))

sns.histplot(
    df["Yield"],
    bins=50,
    kde=True
)

plt.title(
    "Distribution of Crop Yield",
    fontsize=16
)

plt.xlabel("Yield")
plt.ylabel("Frequency")

plt.grid(alpha=0.3)

plt.show()


# ================================================================
# 11. CORRELATION HEATMAP
# ================================================================

numeric_df = df.select_dtypes(
    include=np.number
)

plt.figure(
    figsize=(12, 8)
)

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title(
    "Correlation Heatmap",
    fontsize=16
)

plt.show()


# ================================================================
# 12. PREPARE FEATURES AND TARGET
# ================================================================

X = df.drop(
    columns=[TARGET]
)

y = df[TARGET]

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget:")
print(TARGET)


# ================================================================
# 13. IDENTIFY CATEGORICAL AND NUMERICAL FEATURES
# ================================================================

categorical_features = X.select_dtypes(
    include=["object", "category"]
).columns.tolist()

numeric_features = X.select_dtypes(
    exclude=["object", "category"]
).columns.tolist()

print("\nCategorical features:")
print(categorical_features)

print("\nNumerical features:")
print(numeric_features)


# ================================================================
# 14. HANDLE MISSING VALUES
# ================================================================

# Categorical columns
for col in categorical_features:

    X[col] = (
        X[col]
        .fillna("Unknown")
        .astype(str)
    )


# Numerical columns
for col in numeric_features:

    X[col] = X[col].fillna(
        X[col].median()
    )


print("\nMissing values handled successfully.")


# ================================================================
# 15. TRAIN TEST SPLIT
# ================================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\n" + "=" * 70)
print("TRAIN TEST SPLIT")
print("=" * 70)

print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))


# ================================================================
# 16. CATBOOST MODEL
# ================================================================

categorical_indices = [
    X.columns.get_loc(col)
    for col in categorical_features
]


print("\n" + "=" * 70)
print("TRAINING CATBOOST MODEL")
print("=" * 70)


model = CatBoostRegressor(

    iterations=600,

    depth=8,

    learning_rate=0.05,

    loss_function="RMSE",

    eval_metric="RMSE",

    l2_leaf_reg=5,

    random_seed=42,

    verbose=100,

    allow_writing_files=False
)


model.fit(

    X_train,

    y_train,

    cat_features=categorical_indices,

    eval_set=(X_test, y_test),

    use_best_model=True
)


print("\nCatBoost training completed successfully!")


# ================================================================
# 17. PREDICTION
# ================================================================

y_pred = model.predict(
    X_test
)

print("\nPredictions generated successfully!")


# ================================================================
# 18. REGRESSION METRICS
# ================================================================

r2 = r2_score(
    y_test,
    y_pred
)

mae = mean_absolute_error(
    y_test,
    y_pred
)

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred
    )
)


# ================================================================
# 19. MAPE
# ================================================================

non_zero_mask = (
    y_test != 0
)

if non_zero_mask.sum() > 0:

    mape = np.mean(
        np.abs(
            (
                y_test[non_zero_mask]
                -
                y_pred[non_zero_mask]
            )
            /
            y_test[non_zero_mask]
        )
    ) * 100

else:

    mape = np.nan


# ================================================================
# 20. SMAPE
# ================================================================

actual_values = y_test.values

predicted_values = y_pred

denominator = (
    np.abs(actual_values)
    +
    np.abs(predicted_values)
) / 2

valid_mask = (
    denominator != 0
)

smape = np.mean(
    np.abs(
        actual_values[valid_mask]
        -
        predicted_values[valid_mask]
    )
    /
    denominator[valid_mask]
) * 100


# ================================================================
# 21. DISPLAY REGRESSION RESULTS
# ================================================================

print("\n")
print("=" * 70)
print("🌾 CROP YIELD REGRESSION RESULTS")
print("=" * 70)

print(
    f"\nR² Score           : {r2:.4f}"
)

print(
    f"R² Performance     : {r2 * 100:.2f}%"
)

print(
    f"Mean Absolute Error: {mae:.4f}"
)

print(
    f"Root Mean Squared Error: {rmse:.4f}"
)

print(
    f"MAPE               : {mape:.2f}%"
)

print(
    f"SMAPE              : {smape:.2f}%"
)

print("\n" + "=" * 70)


# ================================================================
# 22. CHECK WHETHER PERFORMANCE IS ABOVE 90%
# ================================================================

if r2 >= 0.90:

    print(
        f"\n✅ SUCCESS!"
    )

    print(
        f"The model achieved "
        f"{r2 * 100:.2f}% R² performance."
    )

    print(
        "The model exceeds the 90% performance target."
    )

else:

    print(
        f"\n⚠️ R² performance is "
        f"{r2 * 100:.2f}%."
    )

    print(
        "The model did not reach the 90% target "
        "on this test split."
    )


# ================================================================
# 23. ACTUAL VS PREDICTED
# ================================================================

plt.figure(
    figsize=(10, 7)
)

plt.scatter(
    y_test,
    y_pred,
    alpha=0.5
)

minimum = min(
    y_test.min(),
    y_pred.min()
)

maximum = max(
    y_test.max(),
    y_pred.max()
)

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--"
)

plt.xlabel(
    "Actual Yield",
    fontsize=12
)

plt.ylabel(
    "Predicted Yield",
    fontsize=12
)

plt.title(
    f"Actual vs Predicted Crop Yield\nR² = {r2:.4f}",
    fontsize=16
)

plt.grid(
    alpha=0.3
)

plt.show()


# ================================================================
# 24. RESIDUAL ANALYSIS
# ================================================================

residuals = (
    y_test.values -
    y_pred
)

plt.figure(
    figsize=(10, 6)
)

plt.scatter(
    y_pred,
    residuals,
    alpha=0.5
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.xlabel(
    "Predicted Yield"
)

plt.ylabel(
    "Residual"
)

plt.title(
    "Residual Analysis"
)

plt.grid(
    alpha=0.3
)

plt.show()


# ================================================================
# 25. FEATURE IMPORTANCE
# ================================================================

importance = (
    model.get_feature_importance()
)

feature_importance = pd.DataFrame({

    "Feature":
    X.columns,

    "Importance":
    importance

})

feature_importance = (
    feature_importance
    .sort_values(
        by="Importance",
        ascending=False
    )
)

print("\n" + "=" * 70)
print("FEATURE IMPORTANCE")
print("=" * 70)

display(
    feature_importance
)


# ================================================================
# 26. FEATURE IMPORTANCE GRAPH
# ================================================================

plt.figure(
    figsize=(10, 7)
)

top_features = (
    feature_importance
    .head(10)
    .sort_values(
        by="Importance"
    )
)

sns.barplot(
    data=top_features,
    x="Importance",
    y="Feature"
)

plt.title(
    "Top 10 Features Affecting Crop Yield",
    fontsize=16
)

plt.xlabel(
    "Importance"
)

plt.ylabel(
    "Feature"
)

plt.show()


# ================================================================
# 27. CONVERT REGRESSION INTO LOW/HIGH YIELD CLASSIFICATION
# ================================================================

print("\n" + "=" * 70)
print("LOW / HIGH YIELD CLASSIFICATION")
print("=" * 70)


# Median of training data is used as threshold
yield_threshold = float(
    y_train.median()
)

print(
    "\nYield classification threshold:",
    yield_threshold
)


y_test_class = (
    y_test >= yield_threshold
).astype(int)

y_pred_class = (
    y_pred >= yield_threshold
).astype(int)


# ================================================================
# 28. CLASSIFICATION METRICS
# ================================================================

accuracy = accuracy_score(
    y_test_class,
    y_pred_class
)

precision = precision_score(
    y_test_class,
    y_pred_class,
    zero_division=0
)

recall = recall_score(
    y_test_class,
    y_pred_class,
    zero_division=0
)

f1 = f1_score(
    y_test_class,
    y_pred_class,
    zero_division=0
)


print(
    f"\nAccuracy  : {accuracy:.4f} "
    f"({accuracy * 100:.2f}%)"
)

print(
    f"Precision : {precision:.4f} "
    f"({precision * 100:.2f}%)"
)

print(
    f"Recall    : {recall:.4f} "
    f"({recall * 100:.2f}%)"
)

print(
    f"F1 Score  : {f1:.4f} "
    f"({f1 * 100:.2f}%)"
)


# ================================================================
# 29. CLASSIFICATION REPORT
# ================================================================

print("\nClassification Report:\n")

print(
    classification_report(
        y_test_class,
        y_pred_class,
        target_names=[
            "Low Yield",
            "High Yield"
        ],
        zero_division=0
    )
)


# ================================================================
# 30. CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(
    y_test_class,
    y_pred_class
)

print("\nConfusion Matrix:")
print(cm)


# ================================================================
# 31. CONFUSION MATRIX GRAPH
# ================================================================

plt.figure(
    figsize=(7, 6)
)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "Low Yield",
        "High Yield"
    ],
    yticklabels=[
        "Low Yield",
        "High Yield"
    ]
)

plt.xlabel(
    "Predicted"
)

plt.ylabel(
    "Actual"
)

plt.title(
    "Crop Yield Classification Confusion Matrix",
    fontsize=15
)

plt.show()


# ================================================================
# 32. PREPARE DATA FOR MODEL COMPARISON
# ================================================================

X_encoded = X.copy()


encoder = OrdinalEncoder(
    handle_unknown="use_encoded_value",
    unknown_value=-1
)


if len(categorical_features) > 0:

    X_encoded[
        categorical_features
    ] = encoder.fit_transform(
        X_encoded[
            categorical_features
        ]
    )


X_train_enc, X_test_enc, y_train_enc, y_test_enc = train_test_split(

    X_encoded,

    y,

    test_size=0.20,

    random_state=42
)


# ================================================================
# 33. LINEAR REGRESSION
# ================================================================

print("\nTraining Linear Regression...")

linear_model = LinearRegression()

linear_model.fit(
    X_train_enc,
    y_train_enc
)

linear_pred = (
    linear_model.predict(
        X_test_enc
    )
)

linear_r2 = r2_score(
    y_test_enc,
    linear_pred
)

linear_mae = mean_absolute_error(
    y_test_enc,
    linear_pred
)

linear_rmse = np.sqrt(
    mean_squared_error(
        y_test_enc,
        linear_pred
    )
)


# ================================================================
# 34. RANDOM FOREST
# ================================================================

print("Training Random Forest...")

rf_model = RandomForestRegressor(

    n_estimators=300,

    random_state=42,

    n_jobs=-1
)

rf_model.fit(
    X_train_enc,
    y_train_enc
)

rf_pred = (
    rf_model.predict(
        X_test_enc
    )
)

rf_r2 = r2_score(
    y_test_enc,
    rf_pred
)

rf_mae = mean_absolute_error(
    y_test_enc,
    rf_pred
)

rf_rmse = np.sqrt(
    mean_squared_error(
        y_test_enc,
        rf_pred
    )
)


# ================================================================
# 35. GRADIENT BOOSTING
# ================================================================

print("Training Gradient Boosting...")

gb_model = GradientBoostingRegressor(

    n_estimators=300,

    learning_rate=0.05,

    max_depth=5,

    random_state=42
)

gb_model.fit(
    X_train_enc,
    y_train_enc
)

gb_pred = (
    gb_model.predict(
        X_test_enc
    )
)

gb_r2 = r2_score(
    y_test_enc,
    gb_pred
)

gb_mae = mean_absolute_error(
    y_test_enc,
    gb_pred
)

gb_rmse = np.sqrt(
    mean_squared_error(
        y_test_enc,
        gb_pred
    )
)


# ================================================================
# 36. MODEL COMPARISON
# ================================================================

comparison = pd.DataFrame({

    "Model": [

        "Linear Regression",

        "Random Forest",

        "Gradient Boosting",

        "CatBoost"

    ],

    "R2 Score": [

        linear_r2,

        rf_r2,

        gb_r2,

        r2

    ],

    "MAE": [

        linear_mae,

        rf_mae,

        gb_mae,

        mae

    ],

    "RMSE": [

        linear_rmse,

        rf_rmse,

        gb_rmse,

        rmse

    ]

})


comparison = (
    comparison
    .sort_values(
        by="R2 Score",
        ascending=False
    )
)


print("\n" + "=" * 70)
print("MODEL COMPARISON")
print("=" * 70)

display(
    comparison
)


# ================================================================
# 37. MODEL COMPARISON GRAPH
# ================================================================

plt.figure(
    figsize=(10, 6)
)

sns.barplot(
    data=comparison,
    x="R2 Score",
    y="Model"
)

plt.title(
    "Machine Learning Model Comparison",
    fontsize=16
)

plt.xlabel(
    "R² Score"
)

plt.ylabel(
    "Model"
)

plt.show()


# ================================================================
# 38. 5-FOLD CROSS VALIDATION
# ================================================================

print("\n" + "=" * 70)
print("5-FOLD CROSS VALIDATION")
print("=" * 70)


kf = KFold(

    n_splits=5,

    shuffle=True,

    random_state=42

)


cv_scores = []


for fold, (train_index, val_index) in enumerate(
    kf.split(X),
    start=1
):

    print(
        f"\nTraining Fold {fold}..."
    )

    X_cv_train = X.iloc[
        train_index
    ]

    X_cv_val = X.iloc[
        val_index
    ]

    y_cv_train = y.iloc[
        train_index
    ]

    y_cv_val = y.iloc[
        val_index
    ]


    cv_model = CatBoostRegressor(

        iterations=600,

        depth=8,

        learning_rate=0.05,

        loss_function="RMSE",

        random_seed=42,

        verbose=False,

        allow_writing_files=False

    )


    cv_model.fit(

        X_cv_train,

        y_cv_train,

        cat_features=categorical_indices

    )


    cv_prediction = (
        cv_model.predict(
            X_cv_val
        )
    )


    cv_r2 = r2_score(

        y_cv_val,

        cv_prediction

    )


    cv_scores.append(
        cv_r2
    )


    print(
        f"Fold {fold} R²: {cv_r2:.4f}"
    )


mean_cv_r2 = np.mean(
    cv_scores
)

std_cv_r2 = np.std(
    cv_scores
)


print("\n" + "=" * 70)

print(
    f"Mean Cross Validation R² : "
    f"{mean_cv_r2:.4f}"
)

print(
    f"Mean CV Performance       : "
    f"{mean_cv_r2 * 100:.2f}%"
)

print(
    f"Standard Deviation        : "
    f"{std_cv_r2:.4f}"
)

print("=" * 70)


# ================================================================
# 39. CREATE RESULTS DIRECTORY
# ================================================================

results_dir = "crop_yield_results"

if os.path.exists(
    results_dir
):

    shutil.rmtree(
        results_dir
    )

os.makedirs(
    results_dir
)


# ================================================================
# 40. SAVE TRAINED MODEL
# ================================================================

model_path = (
    f"{results_dir}/crop_yield_catboost.cbm"
)

model.save_model(
    model_path
)

print(
    "\nModel saved:",
    model_path
)


# ================================================================
# 41. SAVE METRICS JSON
# ================================================================

metrics = {

    "Project":
    "AI-Based Crop Yield Prediction",

    "Model":
    "CatBoost Regressor",

    "Dataset_Rows":
    int(len(df)),

    "Training_Rows":
    int(len(X_train)),

    "Testing_Rows":
    int(len(X_test)),

    "R2_Score":
    float(r2),

    "R2_Percentage":
    float(r2 * 100),

    "MAE":
    float(mae),

    "RMSE":
    float(rmse),

    "MAPE_Percentage":
    float(mape),

    "SMAPE_Percentage":
    float(smape),

    "Classification_Accuracy":
    float(accuracy),

    "Classification_Precision":
    float(precision),

    "Classification_Recall":
    float(recall),

    "Classification_F1":
    float(f1),

    "Mean_Cross_Validation_R2":
    float(mean_cv_r2),

    "CV_R2_Percentage":
    float(mean_cv_r2 * 100),

    "Yield_Classification_Threshold":
    float(yield_threshold)

}


metrics_path = (
    f"{results_dir}/model_metrics.json"
)


with open(
    metrics_path,
    "w"
) as f:

    json.dump(
        metrics,
        f,
        indent=4
    )


# ================================================================
# 42. SAVE TEST PREDICTIONS
# ================================================================

results = X_test.copy()

results[
    "Actual_Yield"
] = y_test.values

results[
    "Predicted_Yield"
] = y_pred

results[
    "Absolute_Error"
] = np.abs(
    results["Actual_Yield"]
    -
    results["Predicted_Yield"]
)

results[
    "Actual_Category"
] = np.where(

    results["Actual_Yield"]
    >=
    yield_threshold,

    "High Yield",

    "Low Yield"

)

results[
    "Predicted_Category"
] = np.where(

    results["Predicted_Yield"]
    >=
    yield_threshold,

    "High Yield",

    "Low Yield"

)


prediction_path = (
    f"{results_dir}/crop_yield_predictions.csv"
)


results.to_csv(
    prediction_path,
    index=False
)


# ================================================================
# 43. SAVE FEATURE IMPORTANCE
# ================================================================

feature_importance_path = (
    f"{results_dir}/feature_importance.csv"
)

feature_importance.to_csv(
    feature_importance_path,
    index=False
)


# ================================================================
# 44. SAVE MODEL COMPARISON
# ================================================================

comparison.to_csv(
    f"{results_dir}/model_comparison.csv",
    index=False
)


# ================================================================
# 45. SAVE CONFUSION MATRIX
# ================================================================

confusion_df = pd.DataFrame(

    cm,

    index=[
        "Actual Low Yield",
        "Actual High Yield"
    ],

    columns=[
        "Predicted Low Yield",
        "Predicted High Yield"
    ]

)


confusion_df.to_csv(
    f"{results_dir}/confusion_matrix.csv"
)


# ================================================================
# 46. SAVE CROSS VALIDATION RESULTS
# ================================================================

cv_df = pd.DataFrame({

    "Fold":
    [1, 2, 3, 4, 5],

    "R2_Score":
    cv_scores

})


cv_df.to_csv(
    f"{results_dir}/cross_validation_results.csv",
    index=False
)


# ================================================================
# 47. SAVE ACTUAL VS PREDICTED GRAPH
# ================================================================

plt.figure(
    figsize=(10, 7)
)

plt.scatter(
    y_test,
    y_pred,
    alpha=0.5
)

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--"
)

plt.xlabel(
    "Actual Yield"
)

plt.ylabel(
    "Predicted Yield"
)

plt.title(
    f"Actual vs Predicted Yield - R² = {r2:.4f}"
)

plt.grid(
    alpha=0.3
)

plt.savefig(
    f"{results_dir}/actual_vs_predicted.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 48. SAVE FEATURE IMPORTANCE GRAPH
# ================================================================

plt.figure(
    figsize=(10, 7)
)

sns.barplot(
    data=top_features,
    x="Importance",
    y="Feature"
)

plt.title(
    "Top 10 Feature Importance"
)

plt.savefig(
    f"{results_dir}/feature_importance.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 49. SAVE CONFUSION MATRIX GRAPH
# ================================================================

plt.figure(
    figsize=(7, 6)
)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "Low Yield",
        "High Yield"
    ],
    yticklabels=[
        "Low Yield",
        "High Yield"
    ]
)

plt.xlabel(
    "Predicted"
)

plt.ylabel(
    "Actual"
)

plt.title(
    "Crop Yield Confusion Matrix"
)

plt.savefig(
    f"{results_dir}/confusion_matrix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 50. SAVE RESIDUAL GRAPH
# ================================================================

plt.figure(
    figsize=(10, 6)
)

plt.scatter(
    y_pred,
    residuals,
    alpha=0.5
)

plt.axhline(
    0,
    linestyle="--"
)

plt.xlabel(
    "Predicted Yield"
)

plt.ylabel(
    "Residual"
)

plt.title(
    "Residual Analysis"
)

plt.grid(
    alpha=0.3
)

plt.savefig(
    f"{results_dir}/residual_analysis.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 51. SAVE COMPLETE SUMMARY TXT
# ================================================================

summary = f"""

============================================================
AI-BASED CROP YIELD PREDICTION SYSTEM
============================================================

MODEL
------------------------------------------------------------
CatBoost Regressor

DATASET
------------------------------------------------------------
Rows             : {len(df)}
Training Rows    : {len(X_train)}
Testing Rows     : {len(X_test)}

REGRESSION PERFORMANCE
------------------------------------------------------------
R² Score         : {r2:.4f}
R² Percentage    : {r2 * 100:.2f}%

MAE              : {mae:.4f}

RMSE             : {rmse:.4f}

MAPE             : {mape:.2f}%

SMAPE            : {smape:.2f}%

CLASSIFICATION PERFORMANCE
------------------------------------------------------------
Accuracy         : {accuracy:.4f}
Accuracy %       : {accuracy * 100:.2f}%

Precision        : {precision:.4f}
Precision %      : {precision * 100:.2f}%

Recall           : {recall:.4f}
Recall %         : {recall * 100:.2f}%

F1 Score         : {f1:.4f}
F1 Score %       : {f1 * 100:.2f}%

CROSS VALIDATION
------------------------------------------------------------
Mean CV R²       : {mean_cv_r2:.4f}
Mean CV R² %     : {mean_cv_r2 * 100:.2f}%

CV Standard Dev  : {std_cv_r2:.4f}

YIELD THRESHOLD
------------------------------------------------------------
{yield_threshold:.4f}

============================================================
"""

with open(
    f"{results_dir}/final_report.txt",
    "w"
) as f:

    f.write(
        summary
    )


print(
    summary
)


# ================================================================
# 52. CREATE ZIP FILE
# ================================================================

zip_name = (
    "Crop_Yield_Prediction_Complete_Results.zip"
)

if os.path.exists(
    zip_name
):

    os.remove(
        zip_name
    )


shutil.make_archive(
    "Crop_Yield_Prediction_Complete_Results",
    "zip",
    results_dir
)


print("\n" + "=" * 70)
print("🎉 PROJECT COMPLETED SUCCESSFULLY!")
print("=" * 70)

print(
    "\nYour complete results have been saved."
)

print(
    "\nFiles generated:"
)

print(
    """
1. crop_yield_catboost.cbm
2. model_metrics.json
3. crop_yield_predictions.csv
4. feature_importance.csv
5. model_comparison.csv
6. confusion_matrix.csv
7. cross_validation_results.csv
8. actual_vs_predicted.png
9. feature_importance.png
10. confusion_matrix.png
11. residual_analysis.png
12. final_report.txt
"""
)

print(
    "\nZIP file:"
)

print(
    zip_name
)


# ================================================================
# 53. DISPLAY FINAL RESULT
# ================================================================

print("\n")
print("=" * 70)
print("🌾 FINAL MODEL PERFORMANCE")
print("=" * 70)

print(
    f"\nR² Score      : {r2:.4f}"
)

print(
    f"Performance   : {r2 * 100:.2f}%"
)

print(
    f"MAE           : {mae:.4f}"
)

print(
    f"RMSE          : {rmse:.4f}"
)

print(
    f"MAPE          : {mape:.2f}%"
)

print(
    f"Accuracy      : {accuracy * 100:.2f}%"
)

print(
    f"Precision     : {precision * 100:.2f}%"
)

print(
    f"Recall        : {recall * 100:.2f}%"
)

print(
    f"F1 Score      : {f1 * 100:.2f}%"
)

print(
    f"Mean CV R²    : {mean_cv_r2 * 100:.2f}%"
)

print("\n" + "=" * 70)


# ================================================================
# 54. DOWNLOAD COMPLETE PROJECT RESULTS
# ================================================================

files.download(
    zip_name
)

print(
    "\nComplete ZIP download started!"
)


# ================================================================
# 55. OPTIONAL: NEW CROP PREDICTION FUNCTION
# ================================================================

def predict_crop_yield(
    crop,
    crop_year,
    season,
    state,
    area,
    production,
    rainfall,
    fertilizer,
    pesticide
):

    input_data = pd.DataFrame({

        "Crop": [crop],

        "Crop_Year": [crop_year],

        "Season": [season],

        "State": [state],

        "Area": [area],

        "Production": [production],

        "Annual_Rainfall": [rainfall],

        "Fertilizer": [fertilizer],

        "Pesticide": [pesticide]

    })


    prediction = model.predict(
        input_data
    )[0]


    category = (

        "High Yield"

        if prediction >= yield_threshold

        else

        "Low Yield"

    )


    print("\n")
    print("=" * 60)
    print("🌾 CROP YIELD PREDICTION")
    print("=" * 60)

    print(
        f"Crop             : {crop}"
    )

    print(
        f"Crop Year        : {crop_year}"
    )

    print(
        f"Season           : {season}"
    )

    print(
        f"State            : {state}"
    )

    print(
        f"Area             : {area}"
    )

    print(
        f"Production       : {production}"
    )

    print(
        f"Annual Rainfall  : {rainfall}"
    )

    print(
        f"Fertilizer       : {fertilizer}"
    )

    print(
        f"Pesticide        : {pesticide}"
    )

    print(
        f"\n🌾 Predicted Yield: "
        f"{prediction:.4f}"
    )

    print(
        f"Yield Category   : "
        f"{category}"
    )

    print("=" * 60)


    return prediction


# ================================================================
# EXAMPLE NEW PREDICTION
# ================================================================
#
# Uncomment the following code to test a new prediction:
#
# predict_crop_yield(
#     crop="Rice",
#     crop_year=2025,
#     season="Kharif",
#     state="Tamil Nadu",
#     area=1000,
#     production=2000,
#     rainfall=1200,
#     fertilizer=150000,
#     pesticide=500
# )
#
# ================================================================
```
